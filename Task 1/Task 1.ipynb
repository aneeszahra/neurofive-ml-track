 Data Story & Initial Findings

The Titanic dataset contains 891 passengers and 12 columns. 
It includes numerical features (Age, Fare, Pclass) and categorical features (Sex, Embarked). 
Missing values exist, especially in Age (177 missing) and Cabin (687 missing). 
Only 342 passengers survived (38%), indicating a potential gender/class bias to explore later. 
Initial inspection shows no major outliers in Fare, but Age distribution is slightly skewed. 
Preprocessing will require handling missing values and encoding categorical variables.
